```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd
import matplotlib.pyplot as plt
```

```python
LOSS_PATH = "models/temp_results/contours/unet_ns3_augmentation/loss_log.npy"
```

```python
loss_dict = np.load(LOSS_PATH, allow_pickle=True).item()
train_loss = loss_dict['train']
val_loss = loss_dict['val']
```

```python
# plot the training and validation loss

plt.figure(figsize=(10, 5))
plt.plot(train_loss, label='Training Loss')
plt.plot(val_loss, label='Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Training and Validation Loss over Epochs')
plt.grid()
plt.legend()
plt.show()
```

# Multigraph

```python
LOSS_PATHS = [
    "models/temp_results/contours/unet_ns3_augmentation/loss_log.npy",
    "models/temp_results/contours/unet_ns3_classic/loss_log.npy"
]

# make 2 plots on the same figure for the 2 loss logs, train and val
fig, ax = plt.subplots(2, figsize=(10, 8))

for LOSS_PATH in LOSS_PATHS:
    loss_dict = np.load(LOSS_PATH, allow_pickle=True).item()
    train_loss = loss_dict['train']
    val_loss = loss_dict['val']
    
    ax[0].plot(train_loss, label=LOSS_PATH.split('/')[-2])
    ax[1].plot(val_loss, label=LOSS_PATH.split('/')[-2])

ax[0].set_title('Training Loss')
ax[0].set_xlabel('Epochs')
ax[0].set_ylabel('Loss')
ax[0].grid()
ax[0].legend()

ax[1].set_title('Validation Loss')
ax[1].set_xlabel('Epochs')
ax[1].set_ylabel('Loss')
ax[1].grid()
ax[1].legend()

plt.tight_layout()
plt.show()
```
